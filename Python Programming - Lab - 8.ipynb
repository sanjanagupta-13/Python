### 05) WAP to Merge two dictionaries. If a key exists in both dictionaries, add their values.

**Sample Input:**
```
Dict1 = {1: 10, 2: 20, 3: 30}
Dict2 = {2: 40, 3: 50, 4: 60}

```
**Sample Output:**
```
{1: 10, 2: 60, 3: 80, 4: 60}
```

````python
merged = {}

dict1 = {}
dict2 = {}

n = int(input("Enter number of elements in dict1: "))
for i in range(n):
    key = int(input("Enter key: "))
    value = int(input("Enter value: "))
    dict1[key] = value

m = int(input("Enter number of elements in dict2: "))
for i in range(m):
    key = int(input("Enter key: "))
    value = int(input("Enter value: "))
    dict2[key] = value

# Merge logic
for key in dict1:
    merged[key] = dict1[key]

for key in dict2:
    if key in merged:
        merged[key] += dict2[key]
    else:
        merged[key] = dict2[key]

print("Merged Dictionary:", merged)
````

### 06) WAP to Sort a dictionary in ascending order of its values.
Hint: in sorted(), use udf for key parameter


**Sample Input:**
```
{'Math': 70, 'Physics': 85, 'Chemistry': 60}

```
**Sample Output:**
```
{'Chemistry': 60, 'Math': 70, 'Physics': 85}
```

````python
def get_value(item):
    return item[1]

d = {}
n = int(input("Enter number of subjects: "))

for i in range(n):
    key = input("Enter subject: ")
    value = int(input("Enter marks: "))
    d[key] = value

print(dict(sorted(d.items(), key=get_value)))
````

### 07) WAP to find the keys having maximum and minimum values in a dictionary.

### 15) WAP to create a dictionary using dictionary comprehension from two lists containing student names and marks. Include only those students in the dictionary whose marks are greater than or equal to 75.

**Sample Input:**
```
students = ['Amit', 'Riya', 'Neha', 'Karan']
marks = [78, 72, 85, 60]
```
**Sample Output:**
```
{'Amit': 78, 'Neha': 85}
```

````python
names = []
marks = []

n = int(input("Enter number of students: "))

for i in range(n):
    names.append(input("Enter student name: "))
    marks.append(int(input("Enter marks: ")))

result = {names[i]: marks[i] for i in range(n) if marks[i] >= 75}

print("Students with marks >= 75:")
print(result)
````
