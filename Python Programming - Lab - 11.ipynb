### Practical 1: Count Lines, Words and Characters

**Problem Definition:**  
Write a Python program to read text from a file and count the total number of lines, words, and characters. The result should be written to an output file.


```python
fp=open("new.txt","r+")
data=fp.read()
print(data)
fp.close()

lines = data.count("\n") + 1
print(f"lines:{lines}")
words = len(data.split())
print(f"words:{words}")
Characters = len(data.replace("\n", "")) 
print(f"Characters:{Characters}")

out=open("output.txt","w")
out.write("lines: " +str(lines)+ "\n")
out.write("words: " +str(words)+ "\n")
out.write("characters: "+str(Characters)+ "\n")

out.close()

    
```


### Practical 3: Student Marks and Average

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Store marks in a list, calculate the average using a user-defined function, and write student details along with average marks to an output file.


```python
def average(marks):
    return sum(marks) / len(marks)

fp = open("new.txt", "r")

for line in fp:
    data=line.split()
    
    roll=data[0]
    name=data[1]
    marks=list(map(int,data[2:]))
    
    avg = average(marks)

    out = open("output.txt", "a")
    out.write(f"roll :{roll},name:{name},marks:{marks},avg:{avg} \n")


fp.close()
out.close()
```


### Practical 4: Maximum and Minimum using Tuple

**Problem Definition:**  
Write a Python program to read numbers from a file, store them in a tuple, and find the maximum and minimum values. Write the result to an output file.


```python
fp=open("new.txt","r")
data=fp.read()
print(data)
fp.close()

store=tuple(map(int,data.replace(",", "").split()))

out=open("output.txt","w")
out.write("max_values :"+str(max(store))+ "\n")
out.write("min_values :"+str(min(store)))
out.close()

```


### Practical 5: Remove Duplicate Words using Set

**Problem Definition:**  
Write a Python program to read words from a file and remove duplicate words using a set. Store the unique words in an output file.


```python
f1 = open("abc.txt", "r")
words= f1.read().split()
unique_words = set(words)

f2 = open("output.txt", "w")
for word in unique_words:
    f2.write(word + " ")

print("Duplicate words removed and stored in output.txt")

f1.close()
f2.close()
```


### Practical 6: Word Frequency using Dictionary

**Problem Definition:**  
Write a Python program to count the frequency of each word in a text file using a dictionary and write the result to an output file.


```python
word_freq={}
f1 = open("abc.txt", "r")
words= f1.read().split()
for word in words:
    word_freq[word] = words.count(word)
f1.close()
f2 = open("output.txt", "a")
for word, freq in word_freq.items():
        f2.write(f"{word} : {freq}\n")
print(" frequency of words stored in output.txt")    
f2.close()
```


### Practical 7: Student Result and Grade Calculation

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Use a user-defined function to determine grade (A/B/C/Fail) and write the student result to an output file.

**Grade Criteria:**  
- Marks ≥ 75 : Grade A  
- Marks ≥ 60 : Grade B  
- Marks ≥ 40 : Grade C  
- Marks < 40 : Fail


```python
def calculate_Grade(marks):
    avg=sum(marks) / len(marks)
    if(avg>=75):return "Grade A"
    elif (75>avg>=60):return "Grade B"
    elif(60>avg>=40):return"Grade C"
    else:return"Fail"

f1 = open("stuinfo.txt", "r")
for line in f1:
    data = line.split()  
    
    roll_no = data[0]
    name = data[1]
    marks = list(map(int, data[2:]))

    Grade= calculate_Grade(marks)

    f2 = open("output.txt", "a")
    f2.write(f"Roll No:{roll_no},Name:{name},Marks:{marks},Grade:{Grade}\n")
    f2.close()

print("Student details and Grade have been written to output.txt")

f1.close()
```
