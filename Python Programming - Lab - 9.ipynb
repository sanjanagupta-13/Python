### 05) Write a function that returns True if the given string is Palindrome or False otherwise.

```python
n = int(input("Enter a number: "))

def palindrome(n):
    temp = n
    rev = 0

    while temp > 0:
        digit=temp % 10
        rev=rev* 10 + digit
        temp=temp // 10

    if n == rev:
        return True
    else:
        return False

print(palindrome(n))
```

### 07) Write a function to calculate the sum of the first element of each tuples inside the list.

```python
def sum_first_elements(tuples_list):
    total = 0
    for t in tuples_list:
        total += t[0]
    return total


n = int(input("Enter number of tuples: "))
lst = []

for i in range(n):
    t = tuple(map(int, input(f"Enter elements of tuple {i+1}: ").split()))
    lst.append(t)

print("Sum of first elements:", sum_first_elements(lst))
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibonacci(n):
   if (n==0):
       return 0
   elif(n==1):
       return 1
   else:
       return fibonacci(n-1)+fibonacci(n-2)
fibonacci(10)    
```

### 12) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
def upperlower(s):
    no_lower=0
    no_upper=0
    for i in s:
        if i.islower():
            no_lower+=1
        else:
            no_upper+=1
 
    print(f"no_upper={no_upper}, no_lower={ no_lower}")
s=input("enter a string:")
upperlower(s)
```
