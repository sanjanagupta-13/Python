### 07) WAP to implement simple calculator which performs (add,sub,mul,div) of two no. based on user input.

```python
n1 = int(input("enter number1: "))
n2 = int(input("enter number2: "))
op = input("enter an operator (+, -, *, /, %): ")

match op:
    case '+':print("addition:", n1 + n2)
    case '-':print("subtraction:", n1 - n2)
    case '*':  print("multiplication:", n1 * n2)
    case '/':
        if n2 != 0:print("division:", n1 / n2)
        else:print("Error: division by zero")
    case '%':print("modulo:", n1 % n2)
    case _: print("Invalid operator")
```

### 08) WAP to read marks of five subjects. Calculate percentage and print class accordingly. 
#### Fail below 35 
#### Pass Class between 35 to 45 
#### Second Class
#### between 45 to 60
#### First Class between 60 to 70
#### Distinction if more than 70

```python
s1 = int(input("enter subject 1 marks: "))
s2 = int(input("enter subject 2 marks: "))
s3 = int(input("enter subject 3 marks: "))
s4 = int(input("enter subject 4 marks: "))
s5 = int(input("enter subject 5 marks: "))

per = (s1 + s2 + s3 + s4 + s5) / 5

print("Percentage:", per)

if per > 70:print("Distinction")
elif per >= 60 and per <= 70:print("First Class")
elif per >= 45 and per < 60:print("Second Class")
elif per >= 35 and per < 45:print("Pass Class")
else:print("Fail")
```

### 10) WAP to calculate electricity bill based on following criteria. Which takes the unit from the user.
#### a. First 1 to 50 units – Rs. 2.60/unit
#### b. Next 50 to 100 units – Rs. 3.25/unit
#### c. Next 100 to 200 units – Rs. 5.26/unit
#### d. above 200 units – Rs. 8.45/unit

```python
units = int(input("Enter total electricity units: "))
bill = 0

if units <= 50:
    bill = units * 2.60

elif units <= 100:
    bill = (50 * 2.60) + (units - 50) * 3.25

elif units <= 200:
    bill = (50 * 2.60) + (50 * 3.25) + (units - 100) * 5.26

else:
    bill = (50 * 2.60) + (50 * 3.25) + (100 * 5.26) + (units - 200) * 8.45

print("Total Electricity Bill = Rs.", bill)
```
