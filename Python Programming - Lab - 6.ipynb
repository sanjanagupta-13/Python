### 01) WAP to find sum of all the elements in a List.

```python
size=int(input("enter a size:"))
list1=[]
sum=0
for i in range(size):
    list1.append(int(input(f"enter a list element [i]:")))
    sum+=list1[i]
print(sum)
```

### 02) WAP to find largest element in a List.

```python
size=int(input("enter a size:"))
list1=[]

for i in range(size):
    list1.append(int(input(f"enter a list element [i]:")))
print(list1)

largest=list1[0]
for i in range(0,size):
    if list1[i]>largest:
        largest=list1[i]
print(largest)
```

### 05) WAP to split the List into two parts and append the first part to the end.

### 06) WAP to interchange the elements on two positions entered by a user.

```python
l=list(map(int,input("enter a list:").split(" ")))
size=len(l)
fpos=int(input("enter a first position:"))
spos=int(input("enter a second position:"))
for i in range(size):
    temp=l[fpos]
    l[fpos]=l[spos]
    l[spos]=temp
print(l)
```
