### 10) WAP to test if the given tuple is distinct or not.

```

```
(1, 2, 3, 4) is distinct? -> True

### 11) WAP to rotate the elements of a tuple to the right by `k` positions.

**Sample Input:**
```
T = (10, 20, 30, 40, 50)
k = 2
```
**Sample Output:**
```
(40, 50, 10, 20, 30)
```

````python
t=tuple(int(i) for i in input("Enter elements").split())
n=len(t)
k=int(input("Enter k:"))
ans=list(t)
for j in range(k):
    temp=ans[-1]
    for i in range(n-1,0,-1):
        ans[i]=ans[i-1]
    ans[0]=temp
ans=tuple(ans)
print(ans)
````

### 14) WAP to create (index, value) pairs for prime numbers.

**Sample Input:**
```
T = (4, 7, 9, 11, 15)
```
**Sample Output:**
```
((1, 7), (3, 11))
```

````python
t=tuple(int(i) for i in input("Enter elements").split())
prime = []
for i in t:
    flg=1
    for j in range(2,i):
        if(i%j==0):
            flg=0
    if flg:
        prime.append(i)
ans = [(i,(t.index(i))) for i in prime]
print(ans)
print(type(ans[0]))
````

### 16) WAP to compute column-wise sum from a tuple of lists.

**Sample Input:**
```
T = ([1, 2, 3], [4, 5, 6], [7, 8, 9])
```

**Sample Output:**
```
12 15 18

```

````python
T = ([1, 2, 3], [4, 5, 6], [7, 8, 9])

rows = len(T)
cols = len(T[0])

for j in range(cols):
    col_sum = 0
    for i in range(rows):
        col_sum += T[i][j]
    print(col_sum, end=" ")
````
